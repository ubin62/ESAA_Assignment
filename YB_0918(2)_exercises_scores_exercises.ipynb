````python
raw_data = {'first_name': ['Jason', 'Molly', 'Tina', 'Jake', 'Amy'],
            'last_name': ['Miller', 'Jacobson', 'Ali', 'Milner', 'Cooze'],
            'female': [0, 1, 1, 0, 1],
            'age': [42, 52, 36, 24, 73],
            'preTestScore': [4, 24, 31, 2, 3],
            'postTestScore': [25, 94, 57, 62, 70]}

df = pd.DataFrame(raw_data)
df
````

딕셔너리 형태의 raw_data를 pd.DataFrame() 함수에 전달하여 판다스 데이터프레임 구조로 변환

### Step 3. Create a Scatterplot of preTestScore and postTestScore, with the size of each point determined by age
#### Hint: Don't forget to place the labels

````python
plt.scatter(df['preTestScore'], df['postTestScore'], s=df['age'])

plt.title('preTestScore vs postTestScore')
plt.xlabel('preTestScore')
plt.ylabel('postTestScore')

plt.show()
````

### Step 4. Create a Scatterplot of preTestScore and postTestScore.
### This time the size should be 4.5 times the postTestScore and the color determined by sex

````python
plt.scatter(df['preTestScore'], df['postTestScore'],
            s=df['postTestScore'] * 4.5,
            c=df['female'],
            cmap='coolwarm')

plt.title('preTestScore vs postTestScore')
plt.xlabel('preTestScore')
plt.ylabel('postTestScore')

plt.show()
````

### BONUS: Create your own question and answer it.

````python
# 문제: 사전 점수 대비 사후 점수의 상승 폭(postTestScore - preTestScore)이 가장 큰 사람은 누구이며, 몇 점이 올랐나요?

# 1. 점수 상승 폭 계산 컬럼 생성
df['score_increase'] = df['postTestScore'] - df['preTestScore']

# 2. 상승 폭이 가장 큰 행(Row) 조회
top_person = df.loc[df['score_increase'].idxmax()]

print(f"가장 점수가 많이 오른 사람: {top_person['first_name']} {top_person['last_name']}")
print(f"점수 상승 폭: {top_person['score_increase']}점")
````

1. 사후 점수에서 사전 점수를 뺀 값으로 score_increase 컬럼을 만들기,

2. idxmax()로 가장 높은 값을 가진 인덱스를 찾고 .loc[]을 사용해 해당 참가자의 이름과 점수 변화량 추출
